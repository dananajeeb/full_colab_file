```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

print("تم تحميل جميع المكتبات بنجاح! 🚀")

# 1. جداول كلمات وأفعال الحظر متعددة المراحل (Blocklists Tables)
# المرحلة الأولى: أفعال وكلمات خلق الذعر والاستعجال
blacklist_stage1_urgency = ['urgent', 'immediately', 'action required', 'suspension', 'final notice', 'restrict', 'blocked']

# المرحلة الثانية: كلمات الطلبات المالية والبيانات الحساسة
blacklist_stage2_financial = ['verify', 'bank', 'password', 'login', 'account', 'ssn', 'wire transfer', 'credit card', 'click here']
```

```python
# 2. بناء محاكاة برمجية لمجموعة البيانات (Enron/Nazario Hybrid) بحجم تمثيلي لقراءة البيانات
# ملاحظة: عند تشغيل الكود على ملف الـ CSV الفعلي الخاص بك، استبدل هذا الجزء بـ: df = pd.read_csv('your_dataset.csv')
print("[+] جاري إعداد وتشكيل مصفوفة البيانات...")
np.random.seed(42)
num_samples = 2000  # حجم تمثيلي للمحاكاة والتجربة

emails = []
labels = []

# توليد رسائل تصيد (تطابق جداول الحظر)
for _ in range(num_samples // 2):
    txt = f"Dear customer, your account faces {np.random.choice(blacklist_stage1_urgency)}. Please {np.random.choice(blacklist_stage2_financial)} your security settings."
    emails.append(txt)
    labels.append(1)

# توليد رسائل شرعية (Enron Corporate Style)
corporate_words = ['meeting', 'project', 'schedule', 'report', 'presentation', 'attached', 'thanks', 'team discussion']
for _ in range(num_samples // 2):
    txt = f"Hi team, regarding the {np.random.choice(corporate_words)} for next week, please find the documents attached."
    emails.append(txt)
    labels.append(0)

df = pd.DataFrame({'email_text': emails, 'label': labels})

print(f"✅ تم توليد وتجهيز مصفوفة البيانات بنجاح! إجمالي الرسائل: {len(df)} رسالة.")
```

```python
# 3. آلية المقارنة النصية متعددة المراحل واستخراج الميزات الإحصائية لجداول الحظر
def extract_multistage_features(text):
    text_lower = text.lower()
    # حساب تكرار مطابقة كلمات المرحلة الأولى والثانية
    score_stage1 = sum(1 for word in blacklist_stage1_urgency if word in text_lower)
    score_stage2 = sum(1 for word in blacklist_stage2_financial if word in text_lower)
    return pd.Series([score_stage1, score_stage2])

print("[+] جاري تشغيل خوارزمية المطابقة النصية متعددة المراحل مع جداول الحظر...")
df[['stage1_score', 'stage2_score']] = df['email_text'].apply(extract_multistage_features)

print(f"✅ تم الانتهاء من استخراج الميزات الإحصائية وإضافتها للمصفوفة بنجاح!")
```

```python
# 4. تحويل النص بالكامل إلى مصفوفة رقمية (TF-IDF Vectorization) لدمجها مع ميزات الجداول
vectorizer = TfidfVectorizer(max_features=500, stop_words='english')
X_tfidf = vectorizer.fit_transform(df['email_text']).toarray()

# دمج مصفوفة النص مع نتائج ميزات المطابقة النصية للمراحل
X_combined = np.hstack((X_tfidf, df[['stage1_score', 'stage2_score']].values))
y = df['label'].values

print(f"✅ تم تحويل النصوص ودمج الميزات بنجاح! حجم مصفوفة البيانات النهائية (X_combined): {X_combined.shape}")
```

```python
TBL = [
    {
        "action": "تحديث",
        "resource": "الحساب",
        "synonyms_action": ["تعديل", "تجديد", "تأكيد"],
        "synonyms_resource": ["كلمة المرور", "البيانات", "البطاقة"]
    },
    {
        "action": "اربح",
        "resource": "جائزة",
        "synonyms_action": ["مبروك", "استلم", "فزت"],
        "synonyms_resource": ["هدية", "أيفون", "مكافأة"]
    }
]

def Compare(word, target, synonyms):
    if word == target or word in synonyms:
        return True
    return False

def Match_Topic(line, n):
    tokens = line.split()
    for i in range(len(tokens)):
        if Compare(tokens[i], n["action"], n["synonyms_action"]):
            for j in range(i + 1, len(tokens)):
                if Compare(tokens[j], n["resource"], n["synonyms_resource"]):
                    return True
    return False

def detect_social_engineering(text_block):
    lines = text_block.strip().split('\n')
    print("--- جاري فحص النص للكشف عن هجمات الهندسة الاجتماعية ---\n")
    attack_found = False

    for l in lines:
        for n in TBL:
            if Match_Topic(l, n):
                print(f"⚠️ [اكتشاف هجوم هندسة اجتماعية!]")
                print(f"   السطر الخبيث المرصود: '{l}'")
                print(f"   الموضوع المتطابق: (الإجراء: {n['action']} -> المورد: {n['resource']})\n")
                attack_found = True

    if not attack_found:
        print("✅ الفحص مكتمل: النص آمن تماماً، ولم يتم العثور على أي هجمات هندسة اجتماعية.")

text_sample = """
عزيزي العميل، يرجى تجديد البيانات الخاصة بك فوراً
قم بدخول الرابط لتفادي تجميد حسابك
"""

detect_social_engineering(text_sample)
```

```python
def get_app_feature(text):
    lines = text.strip().split('\n')
    for l in lines:
        for n in TBL:
            if Match_Topic(l, n):
                return 1
    return 0

df['app_feature'] = df['email_text'].apply(get_app_feature)
X_combined_final = np.hstack((X_tfidf, df[['stage1_score', 'stage2_score', 'app_feature']].values))

X_train, X_test, y_train, y_test = train_test_split(X_combined_final, y, test_size=0.20, random_state=42, stratify=y)

print(f"[+] تم تقسيم البيانات بنجاح: {len(X_train)} للتدريب، {len(X_test)} للاختبار.")
```

```python
# 6. تدريب نموذج التعلم الآلي الكلاسيكي (Random Forest )
print("[+] جاري تدريب نموذج الغابة العشوائية (Random Forest Classifier)...")
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
```

```python
# 7. مرحلة الاختبار والتنبؤ على الـ 20% المتبقية
y_pred = model.predict(X_test)
```

```python
# 8. استخراج وطباعة المخرجات والنتائج الأمنية ومصفوفة الارتباك
print("\n" + "="*50)
print("نتائج تقييم كفاءة نموذج المطابقة النصية متعدد المراحل")
print("="*50)
print(f"دقة النظام الإجمالية (Accuracy): {accuracy_score(y_test, y_pred) * 100:.2f}%")

print("\n--- مصفوفة الارتباك (Confusion Matrix) ---")
cm = confusion_matrix(y_test, y_pred)
print(f"الرسائل الشرعية المصنفة صحيحة (TN): {cm[0][0]}")
print(f"الإنذارات الخاطئة لرسائل شرعية (FP): {cm[0][1]}")
print(f"رسائل التصيد التي أفلتت من الحظر (FN): {cm[1][0]}")
print(f"رسائل التصيد المكتشفة بنجاح (TP): {cm[1][1]}")

print("\n--- تقرير التصنيف التفصيلي (Classification Report) ---")
print(classification_report(y_test, y_pred, target_names=['شرعية (0)', 'تصيد (1)']))
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

np.random.seed(42)
num_samples = 1000

b1_urgency = ['urgent', 'immediately', 'action required', 'suspension', 'final notice', 'restrict']
b2_financial = ['verify', 'bank', 'password', 'login', 'account', 'ssn', 'wire transfer', 'credit card']

emails = []
labels = []

for _ in range(num_samples // 2):
    txt = f"dear user, this is an {np.random.choice(b1_urgency)} message. please {np.random.choice(b2_financial)} your information now."
    emails.append(txt)
    labels.append(1)

corporate_words = ['meeting', 'project', 'schedule', 'report', 'presentation', 'attached', 'thanks', 'team']
for _ in range(num_samples // 2):
    txt = f"hi team, regarding the {np.random.choice(corporate_words)} for next week, please review the latest updates."
    emails.append(txt)
    labels.append(0)

df = pd.DataFrame({'email_text': emails, 'label': labels})

def extract_blocklist_features(text):
    text_lower = text.lower()
    f1 = sum(1 for w in b1_urgency if w in text_lower)
    f2 = sum(1 for w in b2_financial if w in text_lower)
    return pd.Series([f1, f2])

df[['b1_score', 'b2_score']] = df['email_text'].apply(extract_blocklist_features)

vectorizer = TfidfVectorizer(max_features=100)
X_tfidf = vectorizer.fit_transform(df['email_text']).toarray()
X_combined = np.hstack((X_tfidf, df[['b1_score', 'b2_score']].values))
y = df['label'].values

X_train, X_test, y_train, y_test = train_test_split(X_combined, y, test_size=0.2, random_state=42, stratify=y)

rf = RandomForestClassifier(n_estimators=50, random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

svm = SVC(kernel='linear', random_state=42)
svm.fit(X_train, y_train)
y_pred_svm = svm.predict(X_test)

print("="*60)
print("Random Forest Evaluation:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_rf) * 100:.2f}%")
print(f"Confusion Matrix:\n{confusion_matrix(y_test, y_pred_rf)}")
print(classification_report(y_test, y_pred_rf, target_names=['Ham (0)', 'Phishing (1)']))

print("="*60)
print("SVM Evaluation:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_svm) * 100:.2f}%")
print(f"Confusion Matrix:\n{confusion_matrix(y_test, y_pred_svm)}")
print(classification_report(y_test, y_pred_svm, target_names=['Ham (0)', 'Phishing (1)']))
print("="*60)
```
